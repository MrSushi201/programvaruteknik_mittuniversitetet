# 6. Functions and loop

Functions are the primary building blocks of almost every Python program.
They break code into reusable chunks to avoid repetition.
Loops allow programs to execute code repeatedly either for a set number of times or until a specific condition is met.

## 6.1. What is a function, really?

Functions are values.
In Python, functions are first-class values and can be inspected or assigned to variables.

```python
print(len)
print(type(len))
```

If you assign a new value to a built-in function name, the original function is overwritten in that scope.

```python
# len = "I'm not the len you're looking for."
# print(len)
# print(type(len))
```

To restore the original built-in function, use the "del" keyword to remove the custom name assignment.

```python
len = "I'm not the len you're looking for."
del len
print(len("Hello, World!"))
```

How Python executes functions.
Typing just a function's name inspects it;
parantheses () are required to call or execute it.

```python
# Typing just the name does not execute the function
print(len)

# Use parantheses to call the function
print(len())
```

When Python executes num_letters = len("four"), len("four") evaluates to 4, returning 4 to replace the function call before assigning it to num_letters.

Side effects vs. return values.
A side effect occurs when a function changes or affects something outside itself (e.g., printing text to the screen).
Functions like print() display text as a side effect but return None (an instance of NoneType).

```python
return_value = print("What do I return?")
print(return_value)
print(type(return_value))
```

## 6.2. Write your own functions

The anatomy of a function.
A function consists of a function signature and a function body.

```python
def multiply(x, y): # Function signature or header
    # Function body
    product = x * y
    return product
```

### 6.2.1. Function signature

Function signature begins with def, followed by the function name, parameter list in parentheses (x, y), and a colon ":".
Function names must follow standard Python variable naming rules.

### 6.2.2. Function body

Function body is an indented block containing the executable code.
PEP 8 recommends indenting by 4 spaces.

6.2.3. Return statement

Return statement uses the "return" keyword to exit the function and pass back a value.
Code placed after a return statement inside the body will never run.

```python
def multiply(x, y):
    product = x * y
    return product
    print("You can't see me!") # Never executed
```

Function definition order.
Functions must be defined before they are called in a script; otherwise, a NameError is raised.

```python
# Incorrect order raises NameError:
num = multiply(2, 4)

def multiply(x, y):
    product = x * y
    return product
```

```python
# Correct order
def multiply(x, y):
    product = x * y
    return product

num = multiply(2, 4)
print(num)
```

Functions with no return statement.
Functions without an explicit "return" statement automatically return None.

```python
def greet(name):
    print(f"Hello, {name}!")

greet("Dave")

return_value = greet("Dave")
print(return_value)
```

Docstrings and help().
Docstrings are used to document functions using triple-quoted strings (docstrings) directly below the function signature.
These can be inspected using help().

```python
def multiply(x, y):
    """Return the product of two numbers x and y."""
    product = x * y
    return product

print(help(multiply))
```

## 6.3. Challenge: convert temperature

Write a script defining two functions to convert temperatures between Celsius and Fahrenheit.

1. convert_cel_to_far(C) using F = C * 9 / 5 + 32
2. convert_far_to_cel(F) using C = (F - 32) * 5 / 9

```python
def convert_cel_to_far(celsius):
    return celsius * 9 / 5 + 32

def convert_far_to_cel(fahrenheit):
    return (fahrenheit - 32) * 5 / 9

# Prompt user and display converted results rounded to 2 decimal places
fahrenheit_input = float(input("Enter a temperature in degrees F: "))
celsius_result = convert_far_to_cel(fahrenheit_input)
print(f"{fahrenheit_input} degrees F = {celsius_result:.2f} degrees C")


celsius_input = float(input("Enter a temperature in degrees F: "))
fahrenheit_result = convert_far_to_cel(celsius_input)
print(f"{celsius_input} degrees C = {fahrenheit_result:.2f} degrees F")
```

## 6.4. Run in circles

The while loop.
While loop repeats code as long as a test condition evaluates to True.

```python
n = 1
while n < 5:
    print(n)
    n = n + 1
```

Infinite loops.
If the condition never turns False, the loop runs infinitely.
Stop an infinite loop in IDLE by pressing "Ctrl + C" (raises KeyboardInterrupt).

```python
num = float(input("Enter a positive number: "))
while num <= 0:
    print("That's not a positive number!")
    num = float(input("Enter a positive number: "))
```

The for loop.
For loop executes once for every item in a collection.

```python
for letter in "Python":
    print(letter)
```

Pythonic Code.
Using a for loop to iterate over collection is more "Pythonic" (clearer and cleaner) than manually tracking indices with a while loop.

```python
# Less Pythonic while loop equivalent
word = "Python"
index = 0
while index < len(word):
    print(word[index])
    index = index + 1
```

The range() function.
The range() function generates sequences of integers.

- range(n) generates number from 0 up to n - 1
- range(start, stop) generates numbers from start up to stop - 1

```python
for n in range(3):
    print("Python")

for n in range(10, 20):
    print(n * n)
```

```python
# Bill splitting example
amount = float(input("Enter an amount: "))
for num_people in range(2, 6):
    print(f"{num_people} people: ${amount / num_people:,.2f} each")
```

Nested loops.
Nested loops are loops placed inside other loops.

```python
for n in range(1, 4):
    for j in range(4, 7):
        print(f"n = {n} and j = {j}")
```

## 6.5. Challenge: track your investments

Write a script containing a function invest(amount, rate, years) that calculates compound investment growth over time.

```python
def invest(amount, rate, years):
    for year in range(1, years + 1):
        amount = amount * (1 + rate)
        print(f"year {year}: ${amount:,.2f}")

invest(10000, 0.10, 30)
```

Scope resolution and LEGB rule.
Python resolves variable names in a strict order summarized by LEGB acronym.

1. Local (L): Names defined inside the current function
2. Enclosing (E): Names defined in the local scope of outer/enclosing functions
3. Global (G): Top-level names defined in the script module
4. Built-in (B): Pre-assigned names built into Python (e.g., len, abs, round)

```python
# Enclosing scope example
x = 5

def outer_func():
    y = 3
    def inner_func():
        z = x + y # z is local, y is enclosing, and x is global
        return z
    return inner_func()

outer_func()
```

Modifying global variables ( global keyword ).
Attempting to reassign a global variable locally without declaring it raises an UnboundLocalError.

```python
total = 0

def add_to_total(n):
    global total # Comment out to test global
    total = total + n # Raises UnboundedLocalError
    return print(total)

add_to_total(5)
```
