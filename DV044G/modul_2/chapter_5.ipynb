# 5. Numbers and math

Math and computer programming are closely connected because numbers are an integral part of any programming language.
In Python, you do not need to be a math genius to program well - basic algebra is usually sufficient.

## 5.1. Integers and floating-point numbers

Python has three built-in number data types:

- integers
- floating-point numbers
- complex numbers

Integers ( int ).
An integer is a whole number with no decimal places.

```python
print(type(1))
print(int("25"))
```

Interger literals and formatting.
You can write integer literals directly.
To make large integer literals readable, you can use underscores ( _  ) instead of commas.

```python
print(100000)
print(100_000)
```

Arbitrary precision.
Python integers have no upper limit on size; Python handles extremely large whole numbers automatically.
Floating-point numbers ( float ).
A float is a number containing a decimal place.

```python
print(type(1.0))
print(float("25.5"))
```

Float literals and E-notation.
Large or small float literals can be written with standard decimals, digit-separating underscores, or exponential notation ( E-notation ).

```python
print(1000000.0)
print(1_000_000.0)
print(1e6)
print(200000000000000000.0)
print(1e-4)
```

Floating-point limits (inf and -inf).
Unlike integers, floating-point numbers have a maximum size depending on your computer system.
Exceeding the maximum limit evaluates to ( inf ) or ( -inf ), both of which are of type ( float ).

```python
print(2e400)
n = 2e400
print(type(n))
print(-2e400)
```

## 5.2. Arithmetic operators and expressions

Python supports standard mathematical operators for performing calculations.

Addition ( + ) and subtraction ( - ).
Operands can be int or float.
Adding or subtracting two integers returns an int, while any operation involving a float produces a float.

```python
print(1 + 3)
print(1.0 + 2)
print (1 - 1)
print(5.0 - 3)
print(-3)
print(1 - (-3))
```

Note: PEP 8 recommends placing a single space on both sides of arithmetic operators for readability (e.g., 1 + 1 instead of 1+1).

Multiplication ( * ).

```python
print(3 * 3)
print(2 * 8.0)
```

Standard division ( / ).
The ( / ) operator always returns a float, even when dividing integers evenly.

```python
print(9 / 3)
print(5.0 / 2)
print(int(9 / 3))
```

Integer division ( // ).
The ( // ) floor division operator divides numbers and rounds down to the nearest integer.
It returns a float if at least one operand is a float.

```python
print(9 // 3)
print(5 // 2)
print(-3 // 2)
```

Explanation -3 / 2 equals -1.5.
Rounding down to the next integer gives -2.

Division by zero.
Dividing by zero raises a ZeroDivisionError.

```python
print(1 / 0)
```

Exponent ( ** ).
Raises a base to a power.
The operator supports integers, floats, and negative powers.

```python
print(2 ** 3)
print(9 ** 0.5)
print(2 ** -1)
print(2 ** -2)
```

Modulus operator ( % ).
Returns the remainder of division.
Commonly used to determine if a number "n" is even ( n % 2 == 0).

```python
print(5 % 3)
print(20 % 7)
print(16 % 8)
print(1 % 0)
```

Modulus with negative numbers.
Python calculates modulus using the formula:
r = x - (y * (x // y))

```python
print(5 % -3)
print(5 - (-3 * (5 // -3)))

print(-5 % 3)
print(-5 - (3 * (-5 // 3)))

print(-5 % -3)
print(-5 - (-3 * (-5 // -3)))
```

Arithmetic expression and precedence.
Operators follow standard mathematical order of operations.
( ** ) has highest precedence, followed by ( * ), ( / ), ( // ), and ( % ), then ( + ) and ( - ).

```python
print(2 * 3 - 1)
print(4 / 2 + 2 ** 3)
print(-1 + (-3 * 2 + 4))
```

## 5.4. Make Python lie to you

In Python, adding 0.1 + 0.2 produces an unexpected floating-point output.

```python
print(0.1 + 0.2)
print(0.5 / 10 * 2)
```

Why this occurs.
This is a floating-point representation error inherent to modern computer hardware storing numbers in base-2 (binary) rather than base-10 (decimal).

- The decimal fraction 1/10 becomes an infinitely repeating binary fraction
- Because computers have finite memory, floats store binary approximations
- When printing single numbers like 0.1, Python display the shortest decimal representation that shares that binary approximation. However, summing approximated binaries accumulates small discrepancies

## 5.5. Math functions and number methods

Python includes built-in math functions and methods for numerical operations.

The round() function.
The function rounds numbers to a given number of decimal places or to the nearest integer.

```python
print(round(2.3))
print(round(2.7))
print(round(3.14159, 3))
print(round(2.71828, 2))
```

Rounding Ties to Even (IEEE 754 Standard).
When a number ends in .5 (a tie), Python rounds to the nearest even integer.

```python
print(round(2.5))
print(round(3.5))
```

Floating-point representation errors can occasionally affect rounding results (e.g., round(2.675, 2) produces 2.67 instead of 2.68).

The abs() function.
The function returns the absolute value of a number (always positive, preserving data type).

```python
print(abs(3))
print(abs(-5.0))
```

The pow() function.
The function computes powers ( pow(x, y) is equivalent to x ** y ).
It also accepts an optional third argument for modular exponentiation.
pow(x, y, z) computes (x ** y) % z efficiently.

```python
print(pow(2, 3))
print(pow(2, -2))
print(pow(2, 3, 2))
```

Checking integral floats ( .is_integer() ).
Float objects have an .is_integer() method that returns True if the number hs no fractional component.

```python
num = 2.5
print(num.is_integer())

num = 2.0
print(num.is_integer())
```

## 5.6. Print numbers in style

F-strings allow fine-grained formatting specifiers for displaying numbers.

Fixed-point decimal places (:.nf) where n is an integer.

```python
n = 7.125
print(f"The value of n is {n:.2f}")

n = 7.126
print(f"The value of n is {n:.2f}")

n = 1
print(f"The value of n is {n:.2f}")
```

Thousands separator (:,).

```python
n = 1234567890
print(f"The value of n is {n:,}")
```

Combining formatting specifiers (n:,2f).
Useful for formatting monetary values.

```python
balance = 2000.0
spent = 256.35
remaining = balance - spent
print(f"After spending {spent:.2f}, I was left with ${remaining:,.2f}")
```

Percentage formatting (n:%).
This formatting multiplies by 100 and formats as a percentage.

```python
ratio = 0.9
print(f"Over {ratio:.1%} of Pythonists say 'Real Python rocks!")
print(f"Over {ratio:.2%} of Pythonists say 'Real Python rocks!")
```

## 5.7. Complex numbers

Python provides built-in support for complex numbers using j or J for the imaginary unit.

```python
n = 1 + 2j
print(n)
print(n.real)
print(n.imag)
print(n.conjugate())
```

Complex arithmetic.
Complex numbers support ( + ), ( - ), ( * ), ( ** ), and ( / ).

```python
a = 1 + 2j
b = 3 - 4j
print(a - b)
print(a + b)
print(a * b)
print(a / b)
print(a // b)
```

Real and imaginary properties on standard int and float.
Standard integers and floats also possess .real, .imag, and .conjugate() attributes.

```python
x = 42
print(x.real)
print(x.imag)
print(x.conjugate())
```
