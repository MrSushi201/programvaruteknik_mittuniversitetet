# 4. Strings and string methods

## 4.1. What is a string?

A string is a fundamental Python data type used to represent text.
In Python, string values are instances of the <str> class.

```python
print(type('Hello, World!'))

phrase = "Hello, World!"
print(type(phrase))
```

String literals and quotation marks.
String literals can be declared using either single quotes ('...') or double quotes ("...").

```python
string1 = 'Hello, World!'
string2 = "1234"
string3 = "We're #1!"
string4 = 'I said, "Put it over by the llama."'
```

Mixing quotes improperly without escaping causes a SyntaxError because Python thinks the string ends early.

```python
text = "She said, "What time is it?""
print(text)
```

String length (len()).
The len() built-in function returns the number of characters in a string.

```python
print(len("abc"))

letters = "abc"
num_letters = len(letters)
print(num_letters)
```

Multiline strings.
To maintain PEP 8's recommended 79-character line-length limit, long strings can be broken up across multiple lines:

1. Using backslash ( \ )

```python
paragraph = "This planet has - or rather had - a problem, which was \
this: most of the people living on it were unhappy for pretty much \
of the time."
print(paragraph)

long_string = "This multiline string is \
displayed on one line."
print(long_string)
```

2. Using triple quotes (""" or ''') to preserve line breaks and whitespaces.

```python
print(""""
A example of a 
    string that spans across multiple lines
    that also preserves whitespace.
""")
```

Note: Triple-quoted strings placed at the top of a script or function to document code are called docstrings.

## 4.2. Concatenation, indexing, and slicing

String concatenation ( + ).
String concatenation combines two strings without adding automatic spaces.

```python
string1 = "abra"
string2 = "cadabra"
magic_string = string1 + string2
print(magic_string)

first_name = "Santi"
last_name = "Taweesamarn"
full_name = first_name + " " + last_name
print(full_name)
```

String indexing.
Characters are zero-indexed (0 to len(string) - 1).
Negative indices count backward from the end (-1 is the last character).

```python
flavor = "apple pie"
print(flavor[0])
print(flavor[-1])
```

Attempting to access an index outside the string's length raises an IndexError.

```python
print(flavor[99])
```

String slicing ( [start:end] ).
String slicing extracts a substring from "start" up to, but not including, "end".

```python
flavor = "apple pie"
print(flavor[0:3])
print(flavor[:5])
print(flavor[5:])
print(flavor[:])
```

Slicing out of bounds does not raise an error;
it returns available characters or an empty string "".

```python
flavor = "apple pie"
print(flavor[:14])
print(flavor[13:15])
```

Negative slicing.

```python
flavor = "apple pie"
print(flavor[-9:-6])
print(flavor[-9:])
```

String immutability.
Strings cannot be modified in-place.
Attempting item assignment raises a TypeError.

```python
word = "goal"
word[0] = "f"
```

To alter a string, you must create a new one.

```python
word = "goal"
word = "f" + word[1:]
print(word)
```

## 4.3. Manipulate strings with methods

String method are functions called on a specific string object.

Case conversion ( .lower() ) and ( .upper() ).

```python
print("Jean-luc Picard".lower())

loud_voice = "Can you hear me yet?"
print(loud_voice.upper())
```

Removing whitespace ( .rstrip(), .lstrip(), .strip() ).
The methods remove spaces/line feeds from boundaries (middle spaces are untouched).

```python
name = "  Jean-luc Picard  "
print(name.rstrip())
print(name.lstrip())
print(name.strip())
```

Checking prefixes and suffixes using ( .startswith(), .endswith() ).
These methods are case-sensitive and return Boolean values ( True / False ).

```python
starship = "Enterprise"
print(starship.startswith("en"))
print(starship.startswith("En"))
print(starship.endswith("prise"))
```

Reassignment requirements.
Because strings are immutable, methods return a new string rather than changing the original variable.

```python
name = "Picard"
print(name.upper())
print(name)

name = name.upper()
print(name)
```

## 4.4. Interact with user input

The input() function prompts the user and returns their entry as a string.

```python
prompt = "Hey, what's up? "
user_input = input(prompt)
print("You said: " + user_input)

response = input("What should I shout? ")
shouted_response = response.upper()
print("Well, if you insist...", shouted_response)
```

## 4.6. Working with strings and numbers

Arithmetic behavior on numeric strings.
Multiplying a string by an integer repeats the string.

```python
num = "2"
print(num * 2)
```

Adding a string and an integer raises a TypeError.

```python
print("3" + 3)
```

Converting string o number using ( int() and float() ).

```python
print(int("12"))
print(float("12.34"))
print(int("12.34"))
```

Doubling user input example:

```python
num = input("Enter a number to be doubled: ")
doubled_num = int(num) * 2
print("The doubled number is: " + str(doubled_num))
```

Converting objects/numbers to string ( str() ).

```python
num_pancakes = 10
print("I am going to eat " + str(num_pancakes) + " pancakes for breakfast.")

total_pancakes = 10
pancakes_eaten = 3
print("Only " + str(total_pancakes - pancakes_eaten) + " pancakes left for lunch.")
```

## 4.7. Streamline your print statements

Formatting techniques.

```python
name = "Zaphod"
heads = 2
arms = 3

# 1. .format() method
print("{} has {} heads and {} arms.".format(name, heads, arms))

# 2. f-strings (formatted string literals)
print(f"{name} has {heads} heads and {arms} arms.")

# 3. % operator
print("%s has %d heads and %d arms." % (name, heads, arms))
```

## 4.8. Find a string in a string

Finding substrings using ( .find() ).
The method returns the starting index of the first occurrence, or -1 if not found.

```python
phrase = "The surprise is in here somewhere."
print(phrase.find("surprise"))
print(phrase.find("eyeball"))
```

Replacing substrings using ( .replace() ).
The method replaces occurrences of a substring with new.

```python
my_story = "I'm tall"
print(my_story.replace("tall", "short"))

text = "some of the stuff"
new_text = text.replace("some of", "all")
new_text = new_text.replace("stuff", "things")
print(new_text)
```
