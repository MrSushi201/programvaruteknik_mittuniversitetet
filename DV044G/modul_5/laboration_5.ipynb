```python
def load_agents_from_file(file_name: str) -> dict[str, list[str]]:
    """Load and return agent credentials from a file."""

    ''' PSEUDO CODE
    DEFINE a container for agent data
    ENSURE correct file name ending
    DEFINE path to the specified file
    LOAD agent data from CSV file
    PROCESS each agent record according to the assignment requirements
    STORE the processed data in the container
    RETURN loaded agent data
    '''

    # DEFINE a container for agent data
    agents = {}

    # ENSURE correct file name ending
    if not file_name.endswith(".csv"):
        file_name = file_name + ".csv"

    # DEFINE path to the specified file
    file_path = RESOURCES / file_name

    # LOAD agent data from CSV file
    with open(file_path, "r", newline="") as file:
        reader = csv.DictReader(file, delimiter=";")

        # PROCESS each agent record according to the assignment requirements
        for row in reader:
                username = row["username"]
                password_list = [row["decrypted_g"], row["decrypted_vg"]]

                # STORE the processed data in the container
                agents[username] = password_list

    # RETURN loaded agent data
    return agents
```

```python
def update_authentication_statistics(success: bool, file_name: str = "stats.txt") -> None:
    """Update authentication statistics in the specified file."""

    ''' PSEUDO CODE
    DEFINE path to the specified file
    CREATE file with initial values if it does not exist
    LOAD current authentication statistics from file
    UPDATE statistics according to the authentication result
    SAVE updated statistics to file
    '''

    # DEFINE path to the specified file
    file_path = RESOURCES / file_name

    # CREATE file with initial values if it does not exist
    if not file_path.exists():
        with open(file_path, "w") as file:
            file.write("auth_success=0\n")
            file.write("auth_failure=0\n")

    auth_success = 0
    auth_failure = 0

    # LOAD current authentication statistics from file
    with open(file_path, "r") as file:
        for line in file:
            line = line.strip()

            if line.startswith("auth_success="):
                auth_success = int(line.split("=")[1])

            elif line.startswith("auth_failure="):
                auth_failure = int(line.split("=")[1])

    # UPDATE statistics according to the authentication result
    if success:
        auth_success += 1
    else:
        auth_failure += 1

    # SAVE updated statistics to file
    with open(file_path, "w") as file:
        file.write(f"auth_success={auth_success}\n")
        file.write(f"auth_failure={auth_failure}\n")
```

```python
import logging

def log_auth_attempt(formatted_username: str, success: bool, agents: dict) -> None:
    """Logs an authentication attempt to auth_attempts.log."""
    
    log_file_path = RESOURCES / "auth_attempts.log"
    
    # Logging Requirements
    logging.basicConfig(
        filename=str(log_file_path),
        level=logging.INFO,
        encoding="utf-8",
        format="%(asctime)s - %(levelname)s: %(message)s",
        datefmt="%Y-%m-%d %H:%M:%S",
        filemode="a"
    )
    
    # Log Entries Requirements

    # Successful authentication attempts
    if success:
        logging.info(f"{formatted_username} successfully authenticated")
    
    # Failed authentication attempts
    else:
        # If the username exists in agents.csv
        if formatted_username in agents:
            logging.warning(f"{formatted_username} failed authentication")
        
        # If the username does not exist in agents.csv
        else:
            logging.critical(f"Authentication with unknown username: {formatted_username}!")
```
