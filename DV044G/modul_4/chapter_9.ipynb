## 9.1. Tupler - oföränderliga sekvenser

En tupel är en ordnad sekvens av värden där varje element har ett bestämt index som börjar på 0.

Skapa tupler:

- Literal - skrivs med parenteser och kommatecken, t.ex. (1, 2, 3)
- Enstaka element - kräver ett avslutande kommateckan, t.ex. (1, ). Utan kommateckan tolkar Python uttrycket (1) som ett heltalsuttryck
- Konvertering - funktionen tuple() kan skapa en tupel från en itererbar sekvens, t.ex. tuple("Python") blir ("P", "y" "t", "h", "o", "n"). Om man skikcar in ett icke-itererbart objekt (som ett heltal) kastas ett TypeError

Egenskaper:

- Oföränderliga (immutable) - när en tupel väl har skapats kan dess element inte ändras, läggas till eller tas bort. Att försöka ändra ett element (tup = 5) ger TypeError
- Längd, indexering och slicing - fungerar precos som för strängar. len(tup) returnerar antalet element, tup hämtar det första elementet och tup[1:3] hämtar ett delsnitt

```python
tup = ("P", "y", "t", "h", "o", "n", "-", "V", ".", "3")
print(len(tup))
print(tup)
print(tup[0:6])
```
