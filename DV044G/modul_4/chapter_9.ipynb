```python
# Packing och unpacking
"""
Tuple packing: Värden separerade med kommatecken packas automatiskt till en tupel även utan paranteser, t.ex. coordinates = 4.21, 9.29

Tuple unpacking: Element i en tupel kan packas upp till enskilda variabler i en enda rad, t.ex. x, y = coordinates eller name, age = "David", 34.
Antalet variabler måste stämma exakt överens med antalet element, annars uppstår ett ValueError.

Returvärden från funktioner: Tupler används ofta för att returnera flera värden från en funktion.
"""

coordinates = 4.21, 9.29
print(type(coordinates), coordinates)
x, y = coordinates
print(f"x: {x}, y: {y}")

vowels = ("a", "e", "i", "o", "u")
for vowel in vowels:
    print(vowel.upper())

print("o" in vowels)
print("x" in vowels)

def adder_subtracter(num1, num2):
    return (num1 + num2, num1 - num2)

print(adder_subtracter(3, 2))
```

## 9.2. Listor - Föränderliga sekvenser

En lista är en ordnad sekvens av objekt och är föränderlig (mutable), vilket innebär att element kan ändras, bytas ut, läggas till eller tas bort.

Literal: Skrivs med hakparanteser, t.ex. [1, 2, 3].

list() funktion: Används för att skapa en ny lista från en annan itererbar sekvens, t.ex. list("Python") = ["P", "y", "t", "h", "o", "n"].

Metoden .split(): Delar upp en sträng till en lista baserat på en avgränsare, t.ex. "a, b, c".split(", ").

```python
# Skapa listor
"""
Literal: Skrivs med hakparanteser, t.ex. [1, 2, 3].

list() funktion: Används för att skapa en ny lista från en annan itererbar sekvens, t.ex. list("Python") = ["P", "y", "t", "h", "o", "n"].

Metoden .split(): Delar upp en sträng till en lista baserat på en avgränsare, t.ex. "a, b, c".split(", ").
"""

colors = ["red", "yellow", "green", "blue"]
print(type(colors))
print(colors)

my_python = list("Python")
print(type(my_python))
print(my_python)

groceries = "egg, milk, chase"
print(type(groceries))
print(groceries)

groceries_list = groceries.split(", ")
print(type(groceries_list))
print(groceries_list)

print("a;b;c".split(";"))
print("The quick brown fox".split(" "))
```

```python
# Metoder för att ändra listor
lst1 = ["a", "b", 1, 2]
print(lst1)
lst1.append("x")
print(lst1)
lst1.insert(1, "å")
print(lst1)
lst1.extend("Python")
print(lst1)
lst1.pop(1)
print(lst1)
lst1.remove("P")
print(lst1)
lst1.reverse()
print(lst1)
```
