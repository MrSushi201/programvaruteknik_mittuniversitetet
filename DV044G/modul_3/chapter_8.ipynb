## 8.1. Compare values 

Unconditional code executes line-by-line without making decisions.
Conditional logic allows programs to execute different actions based on whether a conditional expression evaluates to True and False.

- Boolean data type ( bool ) - A boolean expression evaluates to either True or False. Note that both boolean literals must start with a capital letter
- Boolean comperators:
  - == (equals to)
  - != (not equals to)
  - < (less than)
  - <= (less than or equal to)
  - \> (greater than)
  - \>= (greater than or equal to)

String comparisons.
Strings are compared lexiographically (dictionary ordering). For example, "apple" < "astrunaut" evaluates to True because "apple" comes before "astronaut" alphabetically.

Object identity vs. value equality. While "==" checks whether two variables have the same value, the "is" keyword tests object identity - checking whether both variables reference the exact same object in memory ( id(a) == id(b) ).

## 8.2. Add some logic

Complex conditions are constructed using logical operators: "and", "or", and "not".

- "and" keyword returns True only if both expressions are tru
- "or" keyword uses inclusive or logic in Python - returns True if the first statement, the second statement, or both statements are true
- "not" keyword reverses the truth value of a single expression (not True becomes False)

Operator precedence. When evaluating compound expressions, Python follows a strict order of precedence ( from highest to lowest ).

1. Comparators ( <, <=, ==, >=, >, != )
2. not
3. and
4. or

## 8.3. Control the flow of your program

Control flow statements allow your code to branch into different execution paths.

if, elif, and else statements:

- if statement consists of the if keyword, a test condition ending with a colon ( : ), and an indented code block. Leaving off the colon causes a SyntaxError
- else statement executes only when the preceding if condition evaluates to False
- elif statement (short for else if) checks additional conditions sequentially. Python evaluates elif clauses in order and executes only the first block that evaluates to True, skipping all remaining elif and else branches

Avoiding deeply nested if statements.
Nesting if statements inside one another creates complex branching paths that increase program complexity and make code hard to read. Nested logic can often be simplified using logical operators (and / or) or boolean flag variables.

## 8.5. Break out of the pattern

You can precisely control loop execution by combining if statements with loop control keywords:

- break - immediately terminates the loop and jumps to the code following the loop block
- continue - skips the rest of the current iteration and advances immediately to the next iteration
- for ... else / while .. else loops - the else blocks attached to a loop executes only if the loop finishes all iterations naturally without encountering a break statement

## 8.6. Recover from errors

Run-time errors are referred to as exceptions. To prevent programs from crashing, Python provides error handling mechanisms.

Common built-in exceptions:

- ValueError - raised when a function receives an argument of correct type but inappropriate value (e.g., int("not a number"))
- TypeError - raised when an operation is applied to an inappropriate type (e.g., "1" + 2)
- NameError - raised when trying to access an undefined variable name
- ZeroDivisionError - raised when dividing or moduloing by zero
- OverflowError - raised when a floating-point calculation exceeds memory limits (e.g., pow(2.0, 1_000_000))

The try ... except ... else ... finally pattern

```python
try:
    number = int(input("Enter an integer: "))
except ValueError:
    print("That was not a valid integer!")
else:
    print("Code ran successfully without errors!")
finally:
    print("Cleanup tasks always execute here.")
```

- try block - encloses code that might raise an exception
- except block - catches specific exceptions and handles them gracefully
- else block - executes code only if the try block ran without throwing any exceptions
- finally block - executes unconditionally whether an exception occurred or not (useful for resource cleanup like closing files or database connections)
