# 7. Finding and fixing code bugs

Chapter 7 focuses on identifying, understanding, and resolving errors in Python programs using IDLE´s built-in debugger.

## 7.1. Understanding the three types of code errors

Before diving into debugging tools, it is important to distinguish between the three main categories of programming mistakes.

1. Syntax Errors:
  - Occur when code violates the formal grammar rules of Python (e.g., leaving off a closing quotation mark or a trailling colon)
  - Caught by IDLE before a script executes; IDLE highlights the offending line in red or displays an alert box
2. Run-Time Errors (Exceptions):
  - Occur while the program is running (e.g., attempting to divide by zero or referencing an undefined variable)
  - Cause Python to raise an exception, halt program execution (crash), and print a traceback in the interactive window
3. Logic Errors (Bugs):
  - Occur when a syntactically valid program runs without crashing but produces incorrect or unintended output
  - Removing logic errors is called debugging, and a debugger is a dedicated tool that allows you to inspect variable values line-by-line to pinpoint where logic breaks down

## 7.2. Navigating IDLE´s debug control window

Opening the debugger:

- In IDLE's interactive window (Python Shell), select Debug -> Debugger from the menu
- When active, the interactive shell prompt displays [DEBUG ON] next to the prompt

Interface overview.
The debug control window contains two main visual panels:

- Stack panel - displays the line of code that is about to be executed (e.g., >'__main__'.<module>(), line 1: for i in range(1, 4): )
- Locals panel - lists all active local variables and their current assigned values, updating dynamically as execution progresses

The 5 debugger control buttons.
Located in the upper-left corner, these buttons control how execution moves through the program:

- Step - executes the current line of code and pauses at the very next line
- Go - executes code continuously until a breakpoint is encountered or the program finishes
- Over - steps over a function call or loop body, evaluating it completely without stepping line-by-line through its internal code
- Out - executes all remaining lines inside the current function or loop body and pauses immediately after exiting
- Quit - stops execution and closes the active debugging session

## 7.3. Working with breakpoints

A breakpoint tells the debugger to pause continuous execution at a specific line of code so you can inspect the current state of variables:

- How to set - right-click (or Ctrl + Click on macOS) on the target line in your script window and select Set Breakpoint. IDLE highlights the line in yellow
- How to clear - right-click on the highlighted line and select Clear Breakpoint
- Workflow - set a breakpoint near suspicious code, press F5 to run the script, click Go to skip straight to the breakpoint, and then click Step to analyze execution line-by-line

## 7.4. The 4-step debugging method

Chapter 7 establishes a structured 4-step framework for squashing bugs:

1. Make a guess about where the bug is located - separate the script into distinct blocks (e.g., global scope vs. function definitions) and isolate the suspicious region
2. Set a breakpoint and inspect the code - trace execution line-by-line with the Step button, keeping close watch on the Locals panel
3. Identify the rror and attempt to fix it - compare expected variable states against actual variable states to isolate the precise logical error
4. Repeat steps 1-3 - re-run the script with the debugger until the code works as expected

## 7.5. Practical example - debugging add_underscores()

The chapter presents a hands-on exercise featuring a buggy function intended to surround each character of a string with underscores.

```python
# The buggy code
def add_underscores(word):
    new_word = "_"
    for i in range(0, len(word)):
        new_word = new_word + word[i] + "_"
    return new_word

phrase = "hello"
print(add_underscores(phrase))
```

## 7.6. Alternative debugging vs. refactoring

Print debugging:

- Adding print(f"i = {i}: new_word = {new_word}") inside loops allows quick inspection without configuring a debugger
- Disadvantages - requires re-running the entire script for every check and requires manually removing print() statements before releasing the final code

Refactoring to Pythonic code:

- Refactoring is the process of rewriting existing code to make it cleaner and easier to read without changing its functional behavior
- Instead of using index variables with range(0, len(word)), a more Pythonic implementation iterates directly over string characters

```python
def add_underscores(word):
    new_word = "_"
    for char in word:
        new_word = new_word + char + "_"
    return new_word

phrase = "siya"
print(add_underscores(phrase))
```
